A simple baseline being used for comparison in this project is the dummy classifier using majority classification. This predicts the majority class for every observation, giving a simple baseline for comparison.

```python
# Import libraries
import pandas as pd
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Load datasets
X_train = pd.read_csv(r"C:\Users\faith\OneDrive\Documents\MDX1_MSc_Health_Data_Science\CT4101_Machine_Learning\Machine_Learning_Portfolio\CT4101-Machine-Learning-Portfolio\data\processed\X_train.csv")
X_val = pd.read_csv(r"C:\Users\faith\OneDrive\Documents\MDX1_MSc_Health_Data_Science\CT4101_Machine_Learning\Machine_Learning_Portfolio\CT4101-Machine-Learning-Portfolio\data\processed\X_val.csv")
y_train = pd.read_csv(r"C:\Users\faith\OneDrive\Documents\MDX1_MSc_Health_Data_Science\CT4101_Machine_Learning\Machine_Learning_Portfolio\CT4101-Machine-Learning-Portfolio\data\processed\y_train.csv").squeeze("columns")
y_val = pd.read_csv(r"C:\Users\faith\OneDrive\Documents\MDX1_MSc_Health_Data_Science\CT4101_Machine_Learning\Machine_Learning_Portfolio\CT4101-Machine-Learning-Portfolio\data\processed\y_val.csv").squeeze("columns")

# Train dummy baseline
baseline_model = DummyClassifier(strategy="most_frequent")
# Fit on training data
baseline_model.fit(X_train, y_train)
# Predict on validation data
y_pred = baseline_model.predict(X_val)
y_prob = baseline_model.predict_proba(X_val)[:, 1]
```

#### Evaluating the Model

```python
results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-score",
        "ROC-AUC"
    ],
    "Score": [
        accuracy_score(y_val, y_pred),
        precision_score(y_val, y_pred, zero_division=0),
        recall_score(y_val, y_pred, zero_division=0),
        f1_score(y_val, y_pred, zero_division=0),
        roc_auc_score(y_val, y_prob)
    ]
})

print(results.round(3).to_string(index=False))
```

The results show a high accuracy, as expected for such a model.
