First, import the libraries and load the datasets.

```python
# Import libraries
import pandas as pd
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Load datasets
X_train = pd.read_csv(r"C:\Users\faith\OneDrive\Documents\MDX1_MSc_Health_Data_Science\CT4101_Machine_Learning\Machine_Learning_Portfolio\CT4101-Machine-Learning-Portfolio\data\processed\X_train.csv")
X_val = pd.read_csv(r"C:\Users\faith\OneDrive\Documents\MDX1_MSc_Health_Data_Science\CT4101_Machine_Learning\Machine_Learning_Portfolio\CT4101-Machine-Learning-Portfolio\data\processed\X_val.csv")
y_train = pd.read_csv(r"C:\Users\faith\OneDrive\Documents\MDX1_MSc_Health_Data_Science\CT4101_Machine_Learning\Machine_Learning_Portfolio\CT4101-Machine-Learning-Portfolio\data\processed\y_train.csv").squeeze("columns")
y_val = pd.read_csv(r"C:\Users\faith\OneDrive\Documents\MDX1_MSc_Health_Data_Science\CT4101_Machine_Learning\Machine_Learning_Portfolio\CT4101-Machine-Learning-Portfolio\data\processed\y_val.csv").squeeze("columns")
```

#### 5.1 Dummy Classifier Model

A simple baseline being used for comparison in this project is the dummy classifier using majority classification. This predicts the majority class for every observation, giving a simple baseline for comparison. The most frequent strategy was used.

```python
# Train dummy baseline
baseline_model = DummyClassifier(strategy="most_frequent")
# Fit on training data
baseline_model.fit(X_train, y_train)
# Predict on validation data
y_pred_baseline = baseline_model.predict(X_val)
y_prob_baseline = baseline_model.predict_proba(X_val)[:, 1]
```

#### 5.2 Logistic Regression

Logistic Regression is fitted using the default regularisation parameter of C=1.0. The maximum number of iterations was increased to 3,000 to give convergence.

```python
from sklearn.linear_model import LogisticRegression

# Train the logistic regression model 
lr_model = LogisticRegression(C=1.0, max_iter=3000)
# Fit on training data
lr_model.fit(X_train, y_train)
# Predict on validation data
y_pred_lr = lr_model.predict(X_val)
y_prob_lr = lr_model.predict_proba(X_val)[:, 1]
```

#### 5.3 Decision Tree

The Decision Tree had a maximum depth of 5, a minimum of 20 observations required to split a node, and a minimum of 10 observations required in each leaf. These settings aim to reduce overfitting.

```python
from sklearn.tree import DecisionTreeClassifier
# Train the decision tree
dt_model = DecisionTreeClassifier(
    max_depth=5,
    min_samples_split=20,
    min_samples_leaf=10,
    random_state=123
)
# Fit on training data
dt_model.fit(X_train, y_train)
# Predict on validation data
y_pred_dt = dt_model.predict(X_val)
y_prob_dt = dt_model.predict_proba(X_val)[:, 1]
```

#### 5.4 Random Forest Classifier

The Random Forest was fitted using its default hyperparameters.

```python
from sklearn.ensemble import RandomForestClassifier

# Train the random forest
rf_model = RandomForestClassifier()
# Fit on the training data
rf_model.fit(X_train, y_train)
# Predict on validation data
y_pred_rf = rf_model.predict(X_val)
y_prob_rf = rf_model.predict_proba(X_val)[:, 1]
```

Note: No hyperparameter tuning has been performed on these models yet.
